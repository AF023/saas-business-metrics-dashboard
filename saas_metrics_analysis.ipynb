```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# 1. Simulação de Dados de Negócio (SaaS / Assinaturas)
np.random.seed(42)
n_customers = 500

customer_ids = [f"CUST-{1000 + i}" for i in range(n_customers)]
signup_dates = [datetime(2025, 1, 1) + timedelta(days=int(np.random.randint(0, 365))) for _ in range(n_customers)]
plans = np.random.choice(['Basic ($29/mo)', 'Pro ($99/mo)', 'Enterprise ($299/mo)'], size=n_customers, p=[0.5, 0.35, 0.15])
mrr_map = {'Basic ($29/mo)': 29, 'Pro ($99/mo)': 99, 'Enterprise ($299/mo)': 299}

data = []
for cid, sdate, plan in zip(customer_ids, signup_dates, plans):
    monthly_price = mrr_map[plan]
    churn_prob = 0.15 if plan == 'Basic ($29/mo)' else (0.08 if plan == 'Pro ($99/mo)' else 0.03)
    is_churned = np.random.choice([True, False], p=[churn_prob, 1 - churn_prob])

    tenure_months = np.random.randint(1, 12) if is_churned else 12
    churn_date = (sdate + timedelta(days=int(tenure_months * 30))) if is_churned else None

    data.append({
        'customer_id': cid,
        'signup_date': sdate.strftime('%Y-%m-%d'),
        'plan': plan,
        'monthly_revenue': monthly_price,
        'is_churned': is_churned,
        'churn_date': churn_date.strftime('%Y-%m-%d') if churn_date else None,
        'tenure_months': tenure_months
    })

df = pd.DataFrame(data)

# 2. Cálculo de Métricas Chave de Negócio (KPIs)
total_mrr = df[~df['is_churned']]['monthly_revenue'].sum()
churn_rate = (df['is_churned'].sum() / len(df)) * 100
arpu = df['monthly_revenue'].mean()
ltv = arpu * (df['tenure_months'].mean())

print("=== EXECUTIVE SAAS METRICS SUMMARY ===")
print(f"💰 Total Active MRR: ${total_mrr:,.2f}")
print(f"📉 Overall Churn Rate: {churn_rate:.2f}%")
print(f"👤 ARPU (Avg Revenue Per User): ${arpu:.2f}")
print(f"📊 Estimated LTV (Lifetime Value): ${ltv:.2f}")

# 3. Exportar base tratada em CSV
df.to_csv("saas_customer_metrics.csv", index=False)
print("\n✅ Clean dataset saved as 'saas_customer_metrics.csv'")
```
